This dataset falls under the domain of **Education**, and more specifically, educational inequality. **Purpose and Objectives**

The dataset highlights how, despite progress in improving access to education, a significant proportion of children — particularly in low-income and conflict-affected regions — still lack learning opportunities. It also captures how education *quality* diverges sharply between well-equipped schools in affluent areas and under-resourced institutions in marginalized settings, with gender norms further compounding the gap in some societies.

Its objective is to give researchers a **quantifiable, country-level view of educational inequality** (via HDI-linked indicators) so that patterns can be tracked over time and compared across regions, rather than only being discussed qualitatively.

**Scope**

The dataset contains historical data on a single indicator of educational inequality ("Inequality in Education") for countries across the globe, **covering the period from 2010 to 2021** (a span of over a decade). The presence of the *"human rights"* tag suggests that the dataset frames educational inequality as a human rights issue and not purely an economic or statistical one. Likewise, the *"time series analysis"* tag indicates that the dataset is intended to support analysis of trends over multiple years rather than a single-year snapshot.

**Importance and Relevance**

Education access and quality are closely tied to broader human development outcomes such as economic mobility, gender equity, and poverty reduction. Having standardized, comparable country-level metrics helps identify which regions or groups are falling behind. Addressing inequality in education is not just a matter of fairness — it is a critical step toward building equitable societies and empowering individuals to contribute meaningfully to their own development and that of their nations.

The dataset is also **multi-year and multi-country**, which gives it enough real-world complexity (missing values, grouped categories) to be genuinely useful for data cleaning and exploratory analysis work.

## Section 3 · Data Cleaning

Data cleaning prepares the raw table for analysis: renaming the yearly columns, correcting data types, handling duplicates and missing values, and reshaping to one row per country and year.

### 3.1 · Renaming Columns

```python
# ==== SECTION 3.1 · rename the yearly columns ====
# 'Inequality in Education (2010)' -> 'Inequality_2010', and so on
years = range(2010, 2022)
rename_map = {f'Inequality in Education ({year})': f'Inequality_{year}' for year in years}

df = df.rename(columns=rename_map)
df.columns
```

**What is done:** A dictionary maps each long yearly column name, such as Inequality in Education (2010), to a short name such as Inequality_2010, and df.rename() applies it; the range of years is stored in years and reused in Section 3.5. The printed df.columns confirms that the twelve yearly columns are now named Inequality_2010 through Inequality_2021.

**Why it is done:** Short names without spaces or parentheses are easier and safer to use in code, and they allow the year to be extracted later by removing the Inequality_ prefix. Renaming is done before any other cleaning so that all following steps use the same names.

**What it means for the dataset:** Twelve of the 17 columns were renamed and the other five (ISO3, Country, Human Development Groups, UNDP Developing Regions, HDI Rank (2021)) were left unchanged. The number of columns is still 17, so no data was lost.

### 3.2 · Correcting Data Types

```python
# ==== SECTION 3.2 · correct the data types ====
# HDI Rank is a whole number that may be missing -> nullable integer
df['HDI Rank (2021)'] = pd.to_numeric(
    df['HDI Rank (2021)'],
    errors='coerce'
).astype('Int64')

# Group and region are categorical columns
df['Human Development Groups'] = (
    df['Human Development Groups'].astype('category')
)

df['UNDP Developing Regions'] = (
    df['UNDP Developing Regions'].astype('category')
)

df.dtypes
```

**What is done:** HDI Rank (2021) is converted with pd.to_numeric() to the nullable integer type Int64, and Human Development Groups and UNDP Developing Regions are converted to the category type. df.dtypes prints the result: two text columns (ISO3 and Country), two category columns, one Int64 column, and twelve float64 columns.

**Why it is done:** HDI Rank is a whole-number rank, so an integer type is more accurate than a float, and the nullable Int64 type allows its empty entries to stay empty. Categorical columns use less memory, make the set of allowed values explicit, and prepare the encoding steps in Section 5.

**What it means for the dataset:** Human Development Groups now has 4 categories (High, Low, Medium, Very High) and UNDP Developing Regions has 6 categories (AS, EAP, ECA, LAC, SA, SSA). The 4 missing HDI ranks are kept as missing (<NA>) instead of being turned into zero.

```python
# ==== SECTION 3.3 · check and remove duplicate rows ====
duplicate_count = df.duplicated().sum()
df = df.drop_duplicates()

print(f"Duplicate rows found and removed: {duplicate_count} | rows remaining: {len(df)}")
```

**What is done:** df.duplicated().sum() counts the rows that are exact copies of an earlier row, and df.drop_duplicates() removes them. One message reports 0 duplicate rows found and 195 rows remaining.

**Why it is done:** Duplicate rows would count the same country twice and bias every average and count. Removing them is a standard data-quality step, even when the check finds none.

**What it means for the dataset:** There are 0 duplicate rows, so no row was removed and 195 rows remain. Each of the 195 rows is a different country.

### 3.5 · Removing Countries Without Inequality Data

```python
# ==== SECTION 3.5 · remove countries with no inequality data in any year ====
# Gaps inside single years stay as NaN (dropping every row with a gap would remove many
# of the 195 countries); only countries with ALL twelve years missing are removed.
year_columns_clean = [f'Inequality_{year}' for year in years]
df = df[df[year_columns_clean].notna().any(axis=1)].reset_index(drop=True)

df.shape
```

**What is done:** The list year_columns_clean is built from years (it is reused in Section 3.7), and every country whose twelve yearly values are all empty is removed with the notna().any(axis=1) filter, and the index is reset. The shape after this step is (182, 17). Countries that have at least one value in some year are kept, even when other years are empty.

**Why it is done:** A country with no inequality value in any year cannot contribute to any year-based analysis, so keeping it would only add empty rows. Countries with only some empty years are kept, because dropping every row with any gap would remove a large share of the 195 countries.

**What it means for the dataset:** 13 countries were removed (Antigua and Barbuda, Brunei, Dominica, Eritrea, Micronesia, Equatorial Guinea, Grenada, Saint Kitts and Nevis, Libya, Liechtenstein, Monaco, Nauru, North Korea), which leaves 182 countries. Of the 44 countries without a UNDP region, 2 were among the removed ones, so 42 unclassified countries remain, and 3 of the 4 countries without a Human Development Group were removed, which leaves 1 (Somalia).

**What is done:** df.info() prints the structure of the cleaned table: the number of rows, the non-null count and data type of every column, and the memory used. It combines the checks of the previous steps in one output instead of repeating the head and tail views of Section 2.

### 3.7 · Reshaping to Country-Year Format

The cleaned dataset has one row per country and separate columns for 2010 to 2021, which is awkward for year-by-year summaries and plots. Reshaping it to one row per country and year keeps all country and yearly information in a layout that groups, years, and trends can be summarized from directly.

```python
# ==== SECTION 3.7 · reshape from wide to country-year format ====
id_columns = [
    'ISO3',
    'Country',
    'Human Development Groups',
    'UNDP Developing Regions',
    'HDI Rank (2021)'
]

df_long = df.melt(
    id_vars=id_columns,
    value_vars=year_columns_clean,
    var_name='Year',
    value_name='Inequality'
)

# 'Inequality_2010' -> 2010 (integer), then order by country and year
df_long['Year'] = df_long['Year'].str.replace('Inequality_', '', regex=False).astype(int)
df_long = df_long.sort_values(['Country', 'Year']).reset_index(drop=True)

print("Country-year dataset shape:", df_long.shape)

display(df_long.head(15))
```

**What is done:** df.melt() converts the twelve yearly columns into two columns, Year and Inequality, while the five identifier columns are repeated for every year. The year text is turned into an integer by removing the Inequality_ prefix, and the rows are sorted by country and year. The printed shape is (2184, 7), and the first 15 rows show all twelve years of Afghanistan followed by the first three years of Albania.

**Why it is done:** Summaries by year and group and time-based variables need one row per observation, and the country-year layout turns 182 country rows into 2,184 rows that follow each country through time. It also makes the previous-year value in Section 4 straightforward to compute.

**What it means for the dataset:** The table has 2,184 rows, which equals 182 countries times 12 years, and 7 columns (five identifiers, Year, and Inequality). 1,979 of the rows have an inequality value and 205 (9.4%) are empty, so the reshape keeps the gaps instead of hiding them.

Data transformation changes the scale or structure of variables. In this section the inequality values are standardized, a time-based variable is created, and each country-year row receives the previous year's inequality value.

### 4.1 · Standardization

```python
# ==== SECTION 4.1 · standardize the inequality values ====
# z = (value - mean) / std, so that Inequality_std has mean 0 and std 1.
scaler = StandardScaler()

df_long['Inequality_std'] = scaler.fit_transform(
    df_long[['Inequality']]
)

df_long[
    ['Country', 'Year', 'Inequality', 'Inequality_std']
].head()
```

**What is done:** A StandardScaler is fitted on the Inequality column of df_long, and the transformed values are stored in the new column Inequality_std with the formula (value - mean) / standard deviation. The first five rows show Afghanistan, whose 2010 value of 42.809 becomes 1.653 and whose 2011 value of 44.823 becomes 1.792. Missing values stay missing after the transformation.

**Why it is done:** Standardization puts inequality on a common scale with mean 0 and standard deviation 1, so that a value can be read as the number of standard deviations from the overall average. The scaler is fitted on every row of the table, so the z-scores describe each value relative to all years together.

**What it means for the dataset:** The scaler learned a mean of 18.931 and a standard deviation of 14.448 from the 1,979 observed values, and the 205 empty rows remain empty. Inequality_std carries exactly the same information as Inequality, only expressed in standard deviations.

### 4.2 · Time-Based Variables

```python
# ==== SECTION 4.2 · time-based variable ====
# Number of years since 2010 (2010 -> 0, 2021 -> 11)
df_long['Year_Since_2010'] = df_long['Year'] - 2010

df_long[
    ['Country', 'Year', 'Year_Since_2010']
].head()
```

**What is done:** The new column Year_Since_2010 is created by subtracting 2010 from Year, so 2010 becomes 0 and 2021 becomes 11. The first five rows show Afghanistan with values 0 to 4 for the years 2010 to 2014.

**Why it is done:** A small counting variable is easier to read and to use in calculations than a four-digit calendar year, and it expresses time as a distance from the first year of the data. It gives a simple way to describe a trend over the twelve years.

**What it means for the dataset:** Year_Since_2010 ranges from 0 to 11 and has 12 distinct values, one per year. Because it is only Year minus 2010, the two columns carry the same information and are perfectly correlated (r = 1.00).

### 4.3 · Previous-Year Value

```python
# ==== SECTION 4.3 · previous-year value ====
# Previous year's year and inequality value (within each country)
df_long['Previous_Year'] = (
    df_long.groupby('ISO3')['Year'].shift(1)
)

df_long['Previous_Inequality'] = (
    df_long.groupby('ISO3')['Inequality'].shift(1)
)

# Keep the previous value only if it is from the previous calendar year
df_long.loc[
    df_long['Previous_Year'] != df_long['Year'] - 1,
    'Previous_Inequality'
] = np.nan

df_long[
    [
        'Country',
        'Year',
        'Inequality',
        'Previous_Inequality'
    ]
].head(15)
```

**What is done:** Grouping by ISO3, the cell shifts Year and Inequality down by one row to get the previous year's values, and a calendar-year check sets any previous value that is not from the adjacent year to NaN. The first 15 rows show Afghanistan with no previous value in 2010 and the 2010 value carried into the 2011 row.

**Why it is done:** The previous-year value lets the notebook measure how much a country's inequality changed from one year to the next, which the level alone cannot show. The calendar-year check makes sure that a value is never taken from a country's non-adjacent year.

**What it means for the dataset:** Previous_Inequality is available in 1,803 of the 2,184 rows, compared with 1,979 rows that have a current value. The first year (2010) can never have a previous value, so 182 rows are lost at that end even before gaps are counted.

## Section 5 · Feature Engineering

Feature engineering converts and combines existing variables into new columns. In this section the categorical columns are encoded and a year-to-year change feature and an interaction feature are derived.

### 5.1 · Encoding Categorical Variables

```python
# ==== SECTION 5.1 · encode the categorical variables ====
# Human Development Groups is ordinal (Low < Medium < High < Very High),
# so it gets an explicit order: Low = 0 ... Very High = 3.
hdg_order = ['Low', 'Medium', 'High', 'Very High']

df_long['Human_Development_Group_Encoded'] = (
    df_long['Human Development Groups'].map(
        {group: i for i, group in enumerate(hdg_order)}
    )
)

# UNDP Developing Regions is nominal, so it is one-hot encoded.
# Any Region_ columns from an earlier run are removed first, so that
# re-running this cell does not create duplicate columns.
df_long = df_long.drop(
    columns=[c for c in df_long.columns if c.startswith('Region_')]
)

region_dummies = pd.get_dummies(
    df_long['UNDP Developing Regions'],
    prefix='Region',
    dtype=int
)

df_long = pd.concat(
    [df_long, region_dummies],
    axis=1
)

df_long[
    [
        'Human Development Groups',
        'Human_Development_Group_Encoded',
        'UNDP Developing Regions'
    ] + list(region_dummies.columns)
].head()
```

**What is done:** Human Development Groups is ordinal-encoded with the order Low = 0, Medium = 1, High = 2, and Very High = 3, and UNDP Developing Regions is one-hot encoded into the six indicator columns Region_AS, Region_EAP, Region_ECA, Region_LAC, Region_SA, and Region_SSA. Any Region_ columns left from an earlier run are removed first, so re-running the cell does not duplicate them. The first five rows show Afghanistan as Low (0) with Region_SA = 1.

**Why it is done:** Numeric codes make the classification columns usable in calculations and correlations, and the encoding should respect the meaning of each column: the group has a natural order, whereas the regions have no ranking. Encoding the regions as separate indicator columns avoids implying that one region is larger than another.

**What it means for the dataset:** The table now has 6 region indicator columns. The 42 countries without a UNDP region (504 country-year rows) receive zeros in all six columns, so they cannot be told apart from one another and look like a seventh, unnamed region, while the one country without a Human Development Group (Somalia, 12 rows) receives NaN in Human_Development_Group_Encoded. These two gaps are shown as an Unclassified group in Section 7.

### 5.2 · Derived and Interaction Features

```python
# ==== SECTION 5.2 · derived and interaction features ====
# Change in inequality from the previous year
# (positive = inequality increased, negative = it decreased)
df_long['Inequality_Change'] = (
    df_long['Inequality']
    - df_long['Previous_Inequality']
)

# Interaction between development group code and current inequality
df_long['DevGroup_x_Inequality'] = (
    df_long['Human_Development_Group_Encoded'].astype(float)
    * df_long['Inequality']
)

df_long[
    [
        'Country',
        'Year',
        'Inequality',
        'Previous_Inequality',
        'Inequality_Change',
        'DevGroup_x_Inequality'
    ]
].head()
```

**What is done:** Inequality_Change is created as the current inequality minus the previous year's inequality, and DevGroup_x_Inequality is created as the encoded development group multiplied by the current inequality. The first five rows show Afghanistan with a change of 2.014 in 2011 and no change (0.000) in 2012 and 2013, while the interaction is 0.0 in every row shown.

**Why it is done:** The change feature shows whether inequality is rising or falling, which the level alone does not show. The interaction feature combines development level and inequality in one number so the two can be examined together.

**What it means for the dataset:** Inequality_Change is available in 1,790 rows, and 1,145 of them show exactly zero change, which reflects that many yearly values are repeated from the year before. The interaction equals 0 for all 352 observed rows of the Low group because Low is coded as 0, so this feature carries no information for the least developed countries (all 352 zero values in the column come from the Low group).

## Section 7 · Exploratory Data Analysis

This section describes the cleaned country-year data with four families of measures: **frequency**, **central tendency**, **dispersion**, and **position**. Each chart shows one idea with its finding in the title, and figures and tables are saved in `/content/derived/eda/` for a later dashboard. The section reads only from `df_long`; countries without a UNDP region or Human Development Group appear as **Unclassified**. Repeated calculations and plotting steps are defined once as helper functions in 7.0.

### 7.0 · Setup

```python
# ==== SECTION 7.0 · EDA setup: style, output folders, working tables, shared helpers ====
import os

# Output folders: figures are saved as PNG, summary tables as CSV
EDA_DIR = '/content/derived/eda'
FIG_DIR = f'{EDA_DIR}/figures'
TAB_DIR = f'{EDA_DIR}/tables'
os.makedirs(FIG_DIR, exist_ok=True)
os.makedirs(TAB_DIR, exist_ok=True)

# One plotting style and palette for the whole section
sns.set_theme(style='whitegrid', context='notebook')
plt.rcParams.update({
    'figure.dpi': 100, 'axes.titlesize': 12, 'axes.titleweight': 'bold',
    'axes.spines.top': False, 'axes.spines.right': False,
})
MAIN_COLOR, ACCENT_COLOR, GREY_COLOR = '#2a6f97', '#c8553d', '#8a8f98'

HDG_ORDER = ['Low', 'Medium', 'High', 'Very High']
REGION_ORDER = ['SSA', 'SA', 'AS', 'LAC', 'EAP', 'ECA', 'Unclassified']
GROUPINGS = [('HDG', HDG_ORDER, 'Human Development Group'),
             ('Region', REGION_ORDER, 'UNDP Developing Region')]

# Working tables, built ONLY from df_long; a missing group or region becomes "Unclassified"
eda = df_long.copy()
eda['HDG'] = eda['Human Development Groups'].astype(object).fillna('Unclassified')
eda['Region'] = eda['UNDP Developing Regions'].astype(object).fillna('Unclassified')
obs = eda.dropna(subset=['Inequality']).copy()   # country-years with an observed value
countries = eda.drop_duplicates('ISO3')          # one row per country

# Dashboard copy of the source table
df_long.to_csv(f'{EDA_DIR}/df_long.csv', index=False)


# ---- helpers used by every cell of Section 7 (defined once here) ----
def save_fig(name):
    plt.savefig(f'{FIG_DIR}/{name}.png', dpi=150, bbox_inches='tight')

def save_table(table, name):
    table.to_csv(f'{TAB_DIR}/{name}.csv')

def count_table(series, order, name):
    """Counts and shares of each category, in a fixed order, saved as CSV."""
    counts = series.value_counts().reindex(order).fillna(0).astype(int)
    table = pd.DataFrame({'Countries': counts,
                          'Share (%)': (counts / counts.sum() * 100).round(1)})
    save_table(table, name)
    return table

def plot_counts(counts, title, xlabel, ylabel, name, figsize=(9, 4.5)):
    """Labelled bar chart; an 'Unclassified' bar is drawn in grey."""
    colors = [GREY_COLOR if label == 'Unclassified' else MAIN_COLOR for label in counts.index]
    fig, ax = plt.subplots(figsize=figsize)
    bars = ax.bar(counts.index.astype(str), counts.values, color=colors)
    ax.bar_label(bars, padding=3)
    ax.margins(y=0.12)
    ax.set(title=title, xlabel=xlabel, ylabel=ylabel)
    plt.tight_layout()
    save_fig(name)
    plt.show()

def rounded_mode(s):
    """Most frequent whole-number value (the smallest one if there is a tie)."""
    return s.round(0).mode().iloc[0]

def count_outliers(s):
    """Number of values outside 1.5 x IQR from the quartiles."""
    q1, q3 = s.quantile(0.25), s.quantile(0.75)
    return int(((s < q1 - 1.5 * (q3 - q1)) | (s > q3 + 1.5 * (q3 - q1))).sum())

def central_stats(s):
    return pd.Series({'Observations': s.count(), 'Mean': s.mean(), 'Median': s.median(),
                      'Mode': rounded_mode(s), 'Mean - Median': s.mean() - s.median()})

def dispersion_stats(s):
    return pd.Series({'Std': s.std(), 'Variance': s.var(), 'Range': s.max() - s.min(),
                      'IQR': s.quantile(0.75) - s.quantile(0.25),
                      'CV (%)': s.std() / s.mean() * 100, 'Skewness': s.skew(),
                      'Outliers': count_outliers(s)})

def stats_by(column, func, order=None):
    """Apply a statistics function to the inequality values of each category."""
    table = obs.groupby(column)['Inequality'].apply(func).unstack()
    return (table.reindex(order) if order else table).round(3)

def group_region_panels(draw, title, ylabel, name, figsize=(13, 4.8)):
    """Two side-by-side panels: Human Development Group and UNDP Region."""
    fig, axes = plt.subplots(1, 2, figsize=figsize, sharey=True)
    for ax, (column, order, label) in zip(axes, GROUPINGS):
        draw(ax, column, order)
        ax.set_xlabel(label)
    axes[0].set_ylabel(ylabel)
    fig.suptitle(title, fontweight='bold')
    plt.tight_layout()
    save_fig(name)
    plt.show()


print("Countries:", len(countries), "| country-years:", len(eda), "| observed values:", len(obs))
print("Saved to:", EDA_DIR)
```

**What is done:** The cell creates the output folders /content/derived/eda/figures and /content/derived/eda/tables, sets one seaborn theme, three colors, and the group orders, and builds the working tables eda, obs, and countries from df_long with missing groups and regions labeled Unclassified. It also defines the helper functions used by every later cell (count_table, plot_counts, stats_by, central_stats, dispersion_stats, group_region_panels, and others), saves df_long as a CSV file, and prints the counts.

**Why it is done:** Defining the shared calculations and plot layouts once keeps the later cells short and makes every chart and table consistent, instead of copying the same code into each cell. Building all tables from df_long alone lets Section 7 run from the saved CSV files, and the Unclassified label keeps the countries without a region or group visible.

**What it means for the dataset:** The working tables contain 182 countries, 2,184 country-year rows, and 1,979 observed inequality values (205 rows are empty). 42 countries are labeled Unclassified for region and 1 for group, and the source table was saved as df_long.csv.

```python
# ==== SECTION 7.2 · Central tendency: overall mean, median and mode ====
values = obs['Inequality']
central_overall = pd.DataFrame({
    'Value': [values.mean(), values.median(), rounded_mode(values), values.mode().iloc[0]],
    'Observations at mode': [np.nan, np.nan,
                             int((values.round(0) == rounded_mode(values)).sum()),
                             int((values == values.mode().iloc[0]).sum())]
}, index=['Mean', 'Median', 'Mode (rounded to whole number)', 'Mode (exact value)']).round(3)
save_table(central_overall, 'central_tendency_overall')

mean_median_gap = values.mean() - values.median()
print(f"Mean - median = {mean_median_gap:.3f} ({mean_median_gap / values.median() * 100:.1f}% of the median)")
print("Countries behind the exact mode:", obs.loc[values == values.mode().iloc[0], 'Country'].unique().tolist())

central_overall
```

**What is done:** The cell computes the mean, the median, the mode of the values rounded to whole numbers, and the mode of the exact values for all observed inequality values, counts the observations at each mode, and saves the table. It also prints the mean-minus-median difference and the country behind the exact mode.

**Why it is done:** The mean, median, and mode describe the typical value in three ways, and comparing the mean with the median is the simplest test of skew. Because the values are continuous, the mode is computed on rounded values, and the exact mode is shown to explain why it is not meaningful on its own.

**What it means for the dataset:** The mean is 18.931 and the median is 14.965, so the mean is 3.966 points (26.5%) higher, which confirms a right skew. ### 7.4 · Position

**7.4a · Quartiles, percentiles, and deciles**

```python
# ==== SECTION 7.4 · Position: quartiles, percentiles and deciles ====
def quantile_table(levels, prefix):
    """Quantiles of all years, 2010 and 2021 side by side."""
    table = pd.DataFrame({
        'All years': obs['Inequality'].quantile(levels),
        '2010': obs.loc[obs['Year'] == 2010, 'Inequality'].quantile(levels),
        '2021': obs.loc[obs['Year'] == 2021, 'Inequality'].quantile(levels),
    }).round(3)
    table.index = [f'{prefix}{round(level * 100 / (10 if prefix == "D" else 1))}' for level in levels]
    return table

position_table = quantile_table([0.05, 0.10, 0.25, 0.50, 0.75, 0.90, 0.95], 'P')
deciles = quantile_table([i / 10 for i in range(1, 10)], 'D')
save_table(position_table, 'position_percentiles')
save_table(deciles, 'position_deciles')

fig, ax = plt.subplots(figsize=(9, 4.8))
ax.plot(deciles.index, deciles['2010'], marker='o', color=GREY_COLOR, label='2010')
ax.plot(deciles.index, deciles['2021'], marker='o', color=MAIN_COLOR, label='2021')
ax.set(title=f'Deciles shift down: the median (D5) moves from {deciles.loc["D5", "2010"]:.1f} '
             f'to {deciles.loc["D5", "2021"]:.1f}',
       xlabel='Decile', ylabel='Inequality in Education')
ax.legend()
plt.tight_layout()
save_fig('7_4_deciles_2010_vs_2021')
plt.show()

display(position_table)
deciles
```

**What is done:** quantile_table() computes the 5th, 10th, 25th, 50th, 75th, 90th, and 95th percentiles and the nine deciles of the inequality values for all years, for 2010, and for 2021, and a line chart compares the deciles of 2010 and 2021. The results are the tables position_table and deciles.

**Why it is done:** Percentiles and deciles show where a given value stands in the distribution and how the whole distribution moves, not just its center. Comparing 2010 with 2021 shows which part of the distribution changed.

**What it means for the dataset:** For all years, the quartiles are 5.904, 14.965, and 29.746. The median falls from 17.825 in 2010 to 12.844 in 2021 and the third decile from 10.496 to 6.533, while the ninth decile hardly moves (42.569 to 42.103), so the improvement happened in the lower and middle parts of the distribution.

**What is done:** The cell ranks the countries with a 2021 value from the most to the least unequal, keeps the ten highest and ten lowest, and draws them as one horizontal bar chart with the highest in red and the lowest in blue. The results are the tables most_unequal and least_unequal.

**7.4c · Change of position, 2010 to 2021**

```python
# ==== SECTION 7.4 · Position: how a country's position changed, 2010 to 2021 ====
# Only countries observed in BOTH years, so the two rankings cover the same set
wide_pos = obs.pivot(index='Country', columns='Year', values='Inequality')[[2010, 2021]].dropna()
wide_pos.columns = ['Value_2010', 'Value_2021']

# Percentile position: 100 = most unequal country of the set
for year in (2010, 2021):
    wide_pos[f'Percentile_{year}'] = wide_pos[f'Value_{year}'].rank(pct=True) * 100
wide_pos['Position_Change'] = wide_pos['Percentile_2021'] - wide_pos['Percentile_2010']
wide_pos = wide_pos.round(2).sort_values('Position_Change')
save_table(wide_pos, 'position_change_2010_2021')

moved_less_5 = (wide_pos['Position_Change'].abs() < 5).mean() * 100
print(f"Countries compared: {len(wide_pos)}")
print(f"Moved less than 5 percentile points: {moved_less_5:.1f}%")
print(f"Correlation of 2010 and 2021 values: {wide_pos['Value_2010'].corr(wide_pos['Value_2021']):.3f}")

movers = pd.concat([wide_pos.head(8), wide_pos.tail(8)])
fig, ax = plt.subplots(figsize=(9, 6))
ax.barh(movers.index, movers['Position_Change'],
        color=[MAIN_COLOR if v < 0 else ACCENT_COLOR for v in movers['Position_Change']])
ax.axvline(0, color='black', linewidth=0.8)
ax.set(title=f'Most countries hold their place: {moved_less_5:.0f}% move less than 5 percentile points',
       xlabel='Change in percentile position, 2010 to 2021\n(negative = less unequal relative to peers)')
plt.tight_layout()
save_fig('7_4_position_change')
plt.show()

display(wide_pos.head(5))
wide_pos.tail(5)
```

**What is done:** The cell selects the countries observed in both 2010 and 2021, converts each year's values into a percentile position (100 is the most unequal country of the set), computes the change in position, and prints the number of countries, the share that moved less than 5 points, and the 2010-2021 correlation. A bar chart shows the eight largest improvements and the eight largest deteriorations.

**Why it is done:** A country can improve in value but still keep its place relative to others, and position measures exactly that. Using only countries observed in both years makes the two rankings comparable.

**What it means for the dataset:** 132 countries are compared, 62.9% of them moved less than 5 percentile points, and the values of the two years are correlated at 0.964, so the order of countries is very stable. The largest improvement in position is Oman (-28.79 points, from 30.54 to 11.95) and the largest deterioration is Burkina Faso (+26.52 points, from 20.97 to 39.15).

### 8.2 · Problem Statement

An international education-development agency has a limited equity budget and must decide which countries to prioritize for education-equity programs. Data from 2010 to 2021 show that educational inequality is persistent, closely tied to human development level, and slowest to improve in the most unequal countries. However, development level alone does not reveal which countries are more unequal than their circumstances would predict, so funding decisions based only on development group risk overlooking them. The problem is the lack of an evidence-based way to identify countries whose educational inequality is higher than expected for their level of development. To address it, we will build a predictive model of 2021 Inequality in Education using HDI rank, Human Development Group, and UNDP region, and evaluate it with RMSE, MAE, and R² against a group-average baseline. Countries whose actual score exceeds the predicted score by more than one RMSE will be flagged as priority candidates for funding review.

**Type: Predictive analytics (regression).**

**Setup:**

## Section 9 · Advanced Analytics: Predictive Model of Educational Inequality

This section implements the advanced analytics chosen in the project plan (Section 8.4): **predictive analytics (regression)**. It builds a model of 2021 Inequality in Education from HDI rank, Human Development Group, and UNDP region, compares it with a group-average baseline, validates it, and uses the gap between actual and predicted values to flag countries that are **more unequal than their development level predicts**. These are the priority candidates for a funding review.

Run Sections 1 to 7 first, because this section reuses `df_long`, `eda`, `obs`, `countries`, `HDG_ORDER`, and the colours defined there. Every figure and table is saved to `/content/derived/model/`. The dashboard recomputes the same model from `df_long.csv`, so its Priority countries tab matches the results below.

### 9.0 · Setup

```python
# ==== SECTION 9.0 · advanced analytics setup: libraries, folders, constants, helper functions ====
from sklearn.linear_model import LinearRegression
from sklearn.ensemble import RandomForestRegressor
from sklearn.model_selection import RepeatedKFold, train_test_split, cross_val_predict

# Output folders for the model: figures as PNG, tables as CSV (same pattern as Section 7)
MODEL_DIR = '/content/derived/model'
MODEL_FIG_DIR = f'{MODEL_DIR}/figures'
MODEL_TAB_DIR = f'{MODEL_DIR}/tables'
os.makedirs(MODEL_FIG_DIR, exist_ok=True)
os.makedirs(MODEL_TAB_DIR, exist_ok=True)

TARGET_YEAR = 2021          # latest year, one row per country
SEED = 42                   # fixed seed so every run gives the same numbers
BASELINE, LINEAR, FOREST = 'Baseline (group mean)', 'Linear Regression', 'Random Forest'
MODEL_NAMES = [BASELINE, LINEAR, FOREST]
MODEL_COLORS = {BASELINE: GREY_COLOR, LINEAR: MAIN_COLOR, FOREST: ACCENT_COLOR}


def regression_metrics(y_true, y_pred):
    """RMSE, MAE and R-squared for one set of predictions."""
    y_true = np.asarray(y_true, dtype=float)
    y_pred = np.asarray(y_pred, dtype=float)
    err = y_true - y_pred
    rmse = float(np.sqrt(np.mean(err ** 2)))
    mae = float(np.mean(np.abs(err)))
    r2 = 1 - float(np.sum(err ** 2)) / float(np.sum((y_true - y_true.mean()) ** 2))
    return rmse, mae, r2


def fit_predict(name, x_train, y_train, g_train, x_test, g_test):
    """Fit one model on the training rows; return predictions for the training and test rows."""
    if name == BASELINE:
        group_means = y_train.groupby(g_train).mean()   # one mean per Human Development Group
        overall = y_train.mean()                        # fallback for an unseen group
        predict = lambda groups: groups.map(group_means).fillna(overall).to_numpy()
        return predict(g_train), predict(g_test)
    if name == LINEAR:
        model = LinearRegression()
    else:
        model = RandomForestRegressor(n_estimators=200, min_samples_leaf=3, random_state=SEED, n_jobs=-1)
    model.fit(x_train, y_train)
    return model.predict(x_train), model.predict(x_test)


def save_model_fig(name):
    plt.savefig(f'{MODEL_FIG_DIR}/{name}.png', dpi=150, bbox_inches='tight')


def save_model_table(table, name):
    table.to_csv(f'{MODEL_TAB_DIR}/{name}.csv', index=False)

print('Advanced analytics setup complete. Outputs go to', MODEL_DIR)
```

**What is done:** The cell imports the model tools from scikit-learn (LinearRegression, RandomForestRegressor, RepeatedKFold, train_test_split, and cross_val_predict), creates the folder `/content/derived/model` with a figures and a tables subfolder, and sets the constants: target year 2021, random seed 42, and the three model names with one colour each. It defines three helpers used by every later cell: `regression_metrics()` for RMSE, MAE, and R², `fit_predict()` which fits one model and returns its training and test predictions, and two save functions.

**Why it is done:** A fixed seed and one set of helpers make every result reproducible and make sure that the baseline and both models are scored by exactly the same code. The baseline is defined inside `fit_predict()`, so its group means are learned only from the training rows, as for any other model.

**What it means for the dataset:** Nothing is computed yet, and the cell prints one confirmation line. It depends on Sections 1 to 7.

### 9.1 · Justification from the EDA

```python
# ==== SECTION 9.1 · justification from the EDA: the evidence behind the technique ====
d21 = eda[(eda['Year'] == TARGET_YEAR)].dropna(subset=['Inequality']).copy()

def variance_explained(frame, column):
    """Share of the variance in inequality explained by the group means of one column (eta squared)."""
    total = ((frame['Inequality'] - frame['Inequality'].mean()) ** 2).sum()
    within = frame.groupby(column)['Inequality'].apply(lambda s: ((s - s.mean()) ** 2).sum()).sum()
    return 1 - within / total

r_prev = df_long[['Inequality', 'Previous_Inequality']].corr().iloc[0, 1]
zero_change = (df_long['Inequality_Change'].dropna() == 0).mean() * 100
max_years = obs.groupby('ISO3')['Inequality'].count().max()

evidence = pd.DataFrame([
    ['Human Development Group explains the spread of 2021 inequality',
     f'{variance_explained(d21, "HDG") * 100:.1f}% of variance', 'Structural features should predict well; the group mean is a fair baseline'],
    ['UNDP region explains the spread of 2021 inequality',
     f'{variance_explained(d21, "Region") * 100:.1f}% of variance', 'Region adds information, so it is a feature (with Unclassified kept)'],
    ['Previous-year value vs current value',
     f'r = {r_prev:.3f}', 'Using it would leak the answer, so it is excluded'],
    ['Year-to-year changes that are exactly zero',
     f'{zero_change:.0f}%', 'Values are carried forward, which rules out time-series forecasting'],
    ['Yearly values per country (maximum)',
     f'{max_years} points', 'Too few points per country to forecast'],
    ['Text field in the data', 'None', 'NLP is not applicable'],
    ['Cost, budget or program-effect columns', 'None', 'No defensible objective or constraints for optimization'],
], columns=['EDA evidence', 'Value', 'Decision it supports'])
save_model_table(evidence, 'justification_evidence')

fig, ax = plt.subplots(figsize=(8, 4))
vals = [variance_explained(d21, 'HDG') * 100, variance_explained(d21, 'Region') * 100]
ax.bar(['Human Development\nGroup', 'UNDP region'], vals, color=[MAIN_COLOR, ACCENT_COLOR])
for i, v in enumerate(vals):
    ax.text(i, v + 1, f'{v:.0f}%', ha='center')
ax.set(title=f'Development group alone explains {vals[0]:.0f}% of the variation in {TARGET_YEAR} inequality',
       ylabel='Share of variance explained (%)', ylim=(0, 100))
plt.tight_layout()
save_model_fig('9_1_variance_explained')
plt.show()

evidence
```

**What is done:** The cell measures, from the 2021 rows, how much of the variation in inequality is explained by Human Development Group and by UNDP region (the share of variance explained by the group means). It recomputes the persistence figures from the EDA (correlation with the previous-year value, share of zero year-to-year changes, and the maximum number of yearly values per country), places each one in a table next to the decision it supports, and draws the two variance shares as a bar chart.

**Why it is done:** The project brief requires the technique to be justified with specific EDA findings and alternatives to be considered. This cell turns each finding into a number that supports one choice: regression on structural features, the exclusion of the previous-year value, and the rejection of the three alternatives.

**What it means for the dataset:** Human Development Group alone explains 65.6% of the variance in 2021 inequality and UNDP region explains 60.4%, so structural features should predict well and the group-mean baseline is a fair and demanding comparison (Finding 2). Region adds information and the unclassified countries are kept as their own category (Finding 4). The previous-year value correlates at r = 0.997 with the current value, 64% of year-to-year changes are exactly zero, and each country has at most 12 yearly values, so it would leak the answer and time-series forecasting has too little to work with (Finding 5). The data have no text field, so NLP does not apply, and no cost, budget, or program-effect column, so there is no defensible objective for optimization. Regression is therefore the technique that the data can support.

### 9.2 · Target, Features, and Leakage Check

```python
# ==== SECTION 9.2 · modelling table: target, features, and the data-leakage check ====
# One row per country for the target year; countries without a value, HDI rank, or group cannot be modelled
candidates = eda[eda['Year'] == TARGET_YEAR]
d = candidates.dropna(subset=['Inequality', 'HDI Rank (2021)', 'Human Development Groups']).reset_index(drop=True)
excluded = sorted(set(countries['Country']) - set(d['Country']))

group_code = {g: i for i, g in enumerate(HDG_ORDER)}          # Low = 0 ... Very High = 3
groups = d['Human Development Groups'].astype(str)
X = pd.DataFrame({
    'HDI_Rank': d['HDI Rank (2021)'].astype(float),
    'HDG_Code': groups.map(group_code).astype(float),
})
X = pd.concat([X, pd.get_dummies(d['Region'], prefix='Region', dtype=int)], axis=1)   # includes Region_Unclassified
y = d['Inequality'].astype(float)

print(f'Modelled countries: {len(d)} | features: {X.shape[1]} | excluded: {len(excluded)}')
print('Excluded (no 2021 value, HDI rank, or group):', ', '.join(excluded))
print('Correlation HDI_Rank vs HDG_Code:', round(X['HDI_Rank'].corr(X['HDG_Code']), 3))

# ---- leakage check: what would happen if the previous-year value were allowed as a feature? ----
prev = candidates.dropna(subset=['Inequality', 'Previous_Inequality'])   # Previous_Inequality comes from Section 4.3
cv_pred = cross_val_predict(LinearRegression(), prev[['Previous_Inequality']], prev['Inequality'], cv=5)
_, _, r2_leak = regression_metrics(prev['Inequality'], cv_pred)

# the same check with the legitimate features, for comparison
cv_pred_ok = cross_val_predict(LinearRegression(), X, y, cv=5)
_, _, r2_ok = regression_metrics(y, cv_pred_ok)

leak_table = pd.DataFrame({
    'Feature set': ['Previous-year inequality only (leakage)', 'HDI rank + group + region (used)'],
    'Cross-validated R²': [round(r2_leak, 3), round(r2_ok, 3)],
})
save_model_table(leak_table, 'leakage_check')

fig, ax = plt.subplots(figsize=(8, 4))
ax.bar(['Previous-year value\n(excluded: leakage)', 'HDI rank, group, region\n(used)'],
       leak_table['Cross-validated R²'], color=[GREY_COLOR, MAIN_COLOR])
for i, v in enumerate(leak_table['Cross-validated R²']):
    ax.text(i, v + 0.02, f'{v:.2f}', ha='center')
ax.set(title=f'The previous-year value would give R² = {r2_leak:.2f}, so it is excluded', ylabel='Cross-validated R²', ylim=(0, 1.1))
plt.tight_layout()
save_model_fig('9_2_leakage_check')
plt.show()

display(X.head())
leak_table
```

**What is done:** The cell builds the modelling table with one row per country for 2021, keeping the countries that have a 2021 value, an HDI rank, and a Human Development Group. The target is Inequality. The nine features are HDI_Rank, HDG_Code (Low = 0 to Very High = 3), and the one-hot UNDP region columns, including Region_Unclassified. It then runs a leakage check: it predicts 2021 inequality from the previous-year value alone and from the chosen features, each with 5-fold cross-validation, and compares the R² of the two.

**Why it is done:** The brief asks for a clear target and a feature selection rationale. The leakage check shows by experiment, and not only by argument, why the strongest predictor in the table is deliberately left out.

**What it means for the dataset:** 176 countries are modelled and 6 are left out: Djibouti, Solomon Islands, Somalia, Syrian Arab Republic, Trinidad and Tobago, and Vanuatu. The previous-year value would give a cross-validated R² of 1.00, because the 2021 values repeat 2020 for every country. That would make the model look perfect and tell the agency nothing, so it is excluded. The chosen features reach R² = 0.71 in the same test. HDI_Rank and HDG_Code correlate at -0.958, so they carry overlapping information, which matters when the importance of features is read in 9.6.

### 9.3 · Baseline and Models on a Hold-Out Split

```python
# ==== SECTION 9.3 · baseline and models on an 80/20 hold-out split: overfitting check ====
idx_train, idx_test = train_test_split(np.arange(len(d)), test_size=0.2, random_state=SEED)

rows = []
for name in MODEL_NAMES:
    p_train, p_test = fit_predict(name, X.iloc[idx_train], y.iloc[idx_train], groups.iloc[idx_train],
                                  X.iloc[idx_test], groups.iloc[idx_test])
    rmse_tr, mae_tr, r2_tr = regression_metrics(y.iloc[idx_train], p_train)
    rmse_te, mae_te, r2_te = regression_metrics(y.iloc[idx_test], p_test)
    rows.append({'Model': name, 'Train RMSE': rmse_tr, 'Test RMSE': rmse_te, 'Test MAE': mae_te,
                 'Train R²': r2_tr, 'Test R²': r2_te, 'Test / Train RMSE': rmse_te / rmse_tr})
holdout = pd.DataFrame(rows).round(3)
save_model_table(holdout, 'holdout_metrics')

fig, ax = plt.subplots(figsize=(9, 4.5))
pos = np.arange(len(holdout))
ax.bar(pos - 0.2, holdout['Train RMSE'], 0.4, label='Train', color=MAIN_COLOR)
ax.bar(pos + 0.2, holdout['Test RMSE'], 0.4, label='Test', color=ACCENT_COLOR)
ax.set_xticks(pos)
ax.set_xticklabels(holdout['Model'])
rf_row = holdout[holdout['Model'] == FOREST].iloc[0]
ax.set(title=f'Random Forest error rises from {rf_row["Train RMSE"]:.2f} (train) to {rf_row["Test RMSE"]:.2f} (test): overfitting',
       ylabel='RMSE')
ax.legend()
plt.tight_layout()
save_model_fig('9_3_holdout_train_vs_test')
plt.show()

holdout
```

**What is done:** The cell splits the 176 countries at random into 140 for training and 36 for testing (80/20, seed 42). The baseline, Linear Regression, and Random Forest are fitted on the training countries only. For each model the cell reports RMSE and R² on the training and test countries, MAE on the test countries, and the ratio of test to training RMSE, and draws the training and test RMSE side by side.

**Why it is done:** Comparing training and test error is the direct check for overfitting that the brief requires. A model that is much better on the countries it has seen than on new ones has memorized them.

**What it means for the dataset:** The Random Forest overfits: its RMSE is 5.36 on the training countries and 7.48 on the test countries, a ratio of 1.40. Linear Regression (7.44 train, 6.92 test) and the baseline (8.43 train, 8.99 test) show no sign of overfitting. The test RMSE of Linear Regression is lower than its training RMSE only by chance, because 36 countries are a small test set. On this split both models beat the baseline, with test R² of 0.82 (Linear Regression) and 0.79 (Random Forest) against 0.69. Because a single small split is unreliable, the next cell uses repeated cross-validation.

### 9.4 · Repeated Cross-Validation

```python
# ==== SECTION 9.4 · repeated 5-fold cross-validation (5 repeats) ====
splitter = RepeatedKFold(n_splits=5, n_repeats=5, random_state=SEED)
fold_scores = {n: {'rmse': [], 'mae': [], 'r2': []} for n in MODEL_NAMES}
oof_sum = {n: np.zeros(len(d)) for n in MODEL_NAMES}     # sum of out-of-fold predictions per country
oof_count = np.zeros(len(d))                             # how many times each country was in a test fold

for idx_tr, idx_te in splitter.split(X):
    oof_count[idx_te] += 1
    for name in MODEL_NAMES:
        _, p_te = fit_predict(name, X.iloc[idx_tr], y.iloc[idx_tr], groups.iloc[idx_tr],
                              X.iloc[idx_te], groups.iloc[idx_te])
        rmse, mae, r2 = regression_metrics(y.iloc[idx_te], p_te)
        fold_scores[name]['rmse'].append(rmse)
        fold_scores[name]['mae'].append(mae)
        fold_scores[name]['r2'].append(r2)
        oof_sum[name][idx_te] += p_te

cv_table = pd.DataFrame([{
    'Model': n,
    'CV RMSE': np.mean(fold_scores[n]['rmse']), 'RMSE std': np.std(fold_scores[n]['rmse']),
    'CV MAE': np.mean(fold_scores[n]['mae']), 'CV R²': np.mean(fold_scores[n]['r2']),
} for n in MODEL_NAMES]).round(3)
save_model_table(cv_table, 'cross_validation_metrics')

fig, ax = plt.subplots(figsize=(9, 4.5))
data = [fold_scores[n]['rmse'] for n in MODEL_NAMES]
bp = ax.boxplot(data, patch_artist=True, tick_labels=MODEL_NAMES)
for patch, n in zip(bp['boxes'], MODEL_NAMES):
    patch.set_facecolor(MODEL_COLORS[n])
best_cv = cv_table[cv_table['Model'] != BASELINE].sort_values('CV RMSE').iloc[0]
base_cv = cv_table[cv_table['Model'] == BASELINE].iloc[0]
ax.set(title=f'{best_cv["Model"]} CV RMSE {best_cv["CV RMSE"]:.2f} vs baseline {base_cv["CV RMSE"]:.2f} (25 folds each)',
       ylabel='RMSE per fold')
plt.tight_layout()
save_model_fig('9_4_cv_rmse_by_model')
plt.show()

cv_table
```

**What is done:** The cell runs 5-fold cross-validation five times with different shuffles, which gives 25 training and test fits per model. All models use the same folds. For every fold the cell stores RMSE, MAE, and R², and it stores each country's prediction whenever the country is in a test fold. The table shows the averages and the standard deviation of RMSE, and the boxplot shows the 25 RMSE values of each model.

**Why it is done:** With only 176 countries, one split can be lucky or unlucky. Repeated cross-validation tests every country several times, gives a stable estimate with a measure of spread, and produces the out-of-fold predictions used in 9.7.

**What it means for the dataset:** The baseline has a cross-validated RMSE of 8.73 (standard deviation 0.88), MAE of 6.69, and R² of 0.62. Linear Regression reaches 7.84 (0.99), 6.13, and 0.687, and Random Forest 7.86 (0.81), 5.93, and 0.691. Both models cut the RMSE by about 10% compared with the baseline. The two models are practically tied: they differ by 0.02 in RMSE, much less than the fold-to-fold standard deviation of about 0.8 to 1.0.

### 9.5 · Model Comparison and Choice

```python
# ==== SECTION 9.5 · model comparison and choice of the final model ====
best_name = best_cv['Model']                 # lowest cross-validated RMSE among the two real models
best_rmse = float(best_cv['CV RMSE'])

comparison = cv_table.copy()
comparison['RMSE vs baseline (%)'] = ((comparison['CV RMSE'] / base_cv['CV RMSE'] - 1) * 100).round(1)
comparison['Folds better than baseline (%)'] = [
    round(np.mean(np.array(fold_scores[n]['rmse']) < np.array(fold_scores[BASELINE]['rmse'])) * 100, 0)
    if n != BASELINE else np.nan for n in comparison['Model']]
save_model_table(comparison, 'model_comparison')

print(f'Chosen model: {best_name} (cross-validated RMSE = {best_rmse:.3f})')
comparison
```

**What is done:** The cell chooses the final model as the real model with the lowest cross-validated RMSE, and compares every model with the baseline in two ways: the percentage change in RMSE, and the share of the 25 folds in which the model beats the baseline on that same fold.

**Why it is done:** The choice of model should follow a stated rule and be tested against the baseline, so that a small average gain is not mistaken for a reliable one.

**What it means for the dataset:** Linear Regression is chosen (cross-validated RMSE 7.835). It lowers the RMSE by 10.2% against the baseline and wins in 76% of the folds, while Random Forest lowers it by 10.0% and wins in 96% of the folds. The two are a statistical tie, and Linear Regression is kept because the rule picks it, it does not overfit (9.3), and its coefficients can be read. The improvement over the baseline is real but modest (R² from 0.62 to 0.69), because the Human Development Group already carries most of the signal. The value of the model is therefore less in its accuracy than in its residuals, which show where a country departs from what its development level predicts.

### 9.6 · Feature Importance

```python
# ==== SECTION 9.6 · feature importance: what drives the predictions ====
rf_full = RandomForestRegressor(n_estimators=300, min_samples_leaf=3, random_state=SEED, n_jobs=-1).fit(X, y)
importance = (pd.DataFrame({'Feature': X.columns, 'Importance': rf_full.feature_importances_})
              .sort_values('Importance').reset_index(drop=True))

lin_full = LinearRegression().fit(X, y)
coefficients = pd.DataFrame({'Feature': X.columns, 'Coefficient': lin_full.coef_.round(3)})
save_model_table(importance.sort_values('Importance', ascending=False), 'feature_importance_random_forest')
save_model_table(coefficients, 'coefficients_linear_regression')

fig, axes = plt.subplots(1, 2, figsize=(12, 4.5))
axes[0].barh(importance['Feature'], importance['Importance'], color=MAIN_COLOR)
top_feat = importance.iloc[-1]
axes[0].set(title=f'{top_feat["Feature"]} carries {top_feat["Importance"] * 100:.0f}% of the importance',
            xlabel='Random Forest importance')
axes[1].barh(coefficients['Feature'], coefficients['Coefficient'],
             color=[ACCENT_COLOR if c > 0 else MAIN_COLOR for c in coefficients['Coefficient']])
axes[1].axvline(0, color='black', linewidth=0.8)
axes[1].set(title='Linear Regression coefficients (points of inequality per unit)', xlabel='Coefficient')
plt.tight_layout()
save_model_fig('9_6_feature_importance')
plt.show()

coefficients
```

**What is done:** The cell fits a Random Forest and a Linear Regression on all 176 countries and shows what each relies on: the forest's feature importance and the regression's coefficients.

**Why it is done:** The brief asks for the important features to be interpreted. Two different models give a cross-check of the same question.

**What it means for the dataset:** HDI_Rank carries 88% of the Random Forest importance. The next largest is Region_AS (Arab States, 2.8%), and HDG_Code carries only 0.7%. This does not mean that the development group is irrelevant: HDI rank and group correlate at -0.958, so the forest splits on the finer HDI rank and the group adds little on top of it. In the regression, each place lower in the HDI ranking adds 0.15 points of inequality (about 15 points over 100 places). The region coefficients are best read relative to each other: South Asia (+9.1) and Arab States (+6.5) are highest and Europe and Central Asia (-9.3) is lowest. These are associations in the data, not causes.

### 9.7 · Predicted vs Actual and Residuals

```python
# ==== SECTION 9.7 · predicted vs actual with out-of-fold predictions, and residual check ====
# Every country is predicted only by models that never saw it
oof_best = oof_sum[best_name] / oof_count

results = pd.DataFrame({
    'Country': d['Country'], 'ISO3': d['ISO3'], 'Human Development Group': groups,
    'UNDP Region': d['Region'], 'HDI Rank (2021)': d['HDI Rank (2021)'].astype(int),
    'Actual': y.round(2), 'Predicted': np.round(oof_best, 2),
})
results['Gap (Actual - Predicted)'] = (results['Actual'] - results['Predicted']).round(2)
rmse_oof, mae_oof, r2_oof = regression_metrics(results['Actual'], results['Predicted'])

fig, axes = plt.subplots(1, 2, figsize=(13, 5))
limit = max(results['Actual'].max(), results['Predicted'].max()) + 3
palette = {'Low': '#c8553d', 'Medium': '#e9a23b', 'High': '#2a6f97', 'Very High': '#16324f'}
for g in HDG_ORDER:
    part = results[results['Human Development Group'] == g]
    axes[0].scatter(part['Predicted'], part['Actual'], s=28, color=palette[g], label=g, alpha=0.85)
axes[0].plot([0, limit], [0, limit], '--', color=GREY_COLOR, label='Actual = Predicted')
axes[0].set(title=f'{best_name}: out-of-fold R² = {r2_oof:.2f}, RMSE = {rmse_oof:.2f}',
            xlabel='Predicted inequality', ylabel='Actual inequality', xlim=(0, limit), ylim=(0, limit))
axes[0].legend(title='Development group')

axes[1].hist(results['Gap (Actual - Predicted)'], bins=20, color=MAIN_COLOR, edgecolor='white')
axes[1].axvline(0, color='black', linewidth=0.8)
axes[1].set(title=f'Gaps are centred near zero (mean {results["Gap (Actual - Predicted)"].mean():.2f}) with a right tail',
            xlabel='Actual minus predicted', ylabel='Number of countries')
plt.tight_layout()
save_model_fig('9_7_predicted_vs_actual')
plt.show()

results.head()
```

It builds the results table with actual value, predicted value, and gap (actual minus predicted) for each country, and plots predicted against actual on the left and the distribution of the gaps on the right.

**Why it is done:** Predictions for countries that were used for training would look too good and would hide the very countries we want to find. Out-of-fold predictions keep the gaps honest, and the plot shows how well the model fits and where it does not.

**What it means for the dataset:** On out-of-fold predictions the model reaches R² = 0.71, RMSE = 7.83, and MAE = 6.10. The gaps have a mean of -0.03 and a standard deviation of 7.86, so they are centred on zero with a right tail of countries above their prediction. It lists the flagged countries with the largest gap first, counts them by region and development group, draws the 15 largest gaps, and shows how the number of flagged countries changes when the threshold is 0.5, 1.0, 1.5, or 2.0 times the RMSE.

**Why it is done:** This is the use of the model that the problem statement asks for: an evidence-based list of countries that are more unequal than their development level explains, which a ranking by development group would overlook.

**What it means for the dataset:** 27 of the 176 countries (15%) are flagged. The largest gaps are Timor-Leste (44.9 actual against 20.5 predicted, +24.4), Bhutan (+17.1), Comoros (+16.4), Turkey (+15.4), and Ivory Coast (+14.0). ### 9.9 · Estimated Impact

```python
# ==== SECTION 9.9 · estimated impact: what closing the gap would mean ====
# If each priority country were brought down to the level its development predicts, how much would inequality fall?
excess = priority['Gap (Actual - Predicted)']
before_high = int((priority['Actual'] >= 30).sum())
after_high = int(((priority['Actual'] - excess) >= 30).sum())
all_actual = results['Actual']
mean_before = all_actual.mean()
mean_after = (all_actual - results['Gap (Actual - Predicted)'].where(results['Priority'], 0)).mean()

impact = pd.DataFrame({
    'Measure': ['Priority countries', 'Average excess over prediction (points)', 'Largest excess (points)',
                'Priority countries scoring 30+ today', 'Of those, still 30+ at predicted level',
                f'Average {TARGET_YEAR} score, all modelled countries (today)', 'Average score if the gap were closed'],
    'Value': [len(priority), round(excess.mean(), 1), round(excess.max(), 1), before_high, after_high,
              round(mean_before, 2), round(mean_after, 2)],
})
save_model_table(impact, 'estimated_impact')

fig, ax = plt.subplots(figsize=(8, 4))
ax.bar(['Today', 'If priority gaps\nwere closed'], [mean_before, mean_after], color=[GREY_COLOR, MAIN_COLOR])
for i, v in enumerate([mean_before, mean_after]):
    ax.text(i, v + 0.3, f'{v:.1f}', ha='center')
ax.set(title=f'Closing the priority gaps would lower the average score by {mean_before - mean_after:.1f} points',
       ylabel=f'Average inequality, {TARGET_YEAR}', ylim=(0, mean_before + 4))
plt.tight_layout()
save_model_fig('9_9_estimated_impact')
plt.show()

impact
```

**What is done:** The cell estimates the size of what is at stake by asking what would change if every flagged country fell to the level its development predicts. It reports the average and largest excess, how many flagged countries score 30 or more today and at the predicted level, and the average 2021 score of the 176 modelled countries before and after.

**Why it is done:** The brief asks for recommendations with expected impact. The data contain no program costs or effects, so the estimate does not claim what a program would achieve. It measures the size of the unexplained gap, which is the most the model suggests could be gained.

**What it means for the dataset:** The flagged countries are on average 12.1 points above their prediction, and the largest excess is 24.4 points. 17 of the 27 score 30 or more today, and 9 would still score 30 or more at the predicted level, so 8 would fall below 30. If every flagged country moved to its predicted level, the average score of the 176 countries would fall from 18.01 to 16.15, a drop of 1.86 points or about 10%.

### 9.11 · What the Model Establishes

**Justification.** Development group explains 65.6% and region 60.4% of the variance in 2021 inequality, the previous-year value is a perfect predictor of 2021 (R² = 1.00) and so would be leakage, and the data have too few yearly points, no text, and no cost data. Regression on structural features is the technique the data support.

**Evaluation.** With repeated 5-fold cross-validation, the baseline has RMSE 8.73 and R² 0.62, Linear Regression 7.84 and 0.687, and Random Forest 7.86 and 0.691. The two models are tied and about 10% better than the baseline. The Random Forest overfits (train RMSE 5.36, test 7.48), and Linear Regression does not. HDI rank carries most of the importance (88% in the forest).

**Interpretation.** The model is accurate enough to describe what development level explains, but its main use is the residual: 27 of 176 countries sit more than one RMSE (7.84 points) above their prediction, led by Timor-Leste, Bhutan, Comoros, and Ivory Coast. Sub-Saharan Africa has the most flagged countries (9).

**Recommendations.**
1. Start the funding review with flagged countries that also have a high actual score: Timor-Leste, Bhutan, Comoros, Ivory Coast, Guinea, Egypt, Morocco, Mauritania, and Papua New Guinea all score 35 or more.
2. Use two lists together: the ranking by absolute level (Low development group, South Asia, and Sub-Saharan Africa) and this model's list of unexpected gaps. Review countries that appear on both first.
3. Before any funding decision, screen out flagged countries with a low actual score. Eight of the 27 score below 20 today, and Turkey, Singapore, and Montenegro are flagged only because the linear model predicts below zero. A minimum score of 20 leaves 19 countries.
4. Set the threshold from the budget: 1.5 times the RMSE gives 12 countries for a small budget, and 0.5 gives 50 for a wide screening.
5. Expected impact: closing the gaps of the flagged countries would remove about 12 points on average from each, move 8 of the 17 high-inequality flagged countries below 30, and lower the average score of all modelled countries by 1.9 points (18.01 to 16.15). This is an upper bound from the data, not a forecast of what a program will achieve.

**Limitations.** The model uses 176 countries and one year, so the estimates are uncertain (the standard deviation of RMSE across folds is about 0.8 to 1.0). It uses only structural features, so a gap may reflect left-out causes such as conflict, income, or school quality, and it shows association, not cause. Many yearly values are carried forward and 2021 repeats 2020, so the target is persistent by construction. The linear model can predict below zero. The two models are statistically tied, and HDI rank and group overlap. Six countries could not be modelled, and the 42 unclassified countries are almost all Very High development, so their results rest on a different group than the rest of the world.
